Simple Linear Regression

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
## Read the Dataset
df=pd.read_csv('height-weight.csv')
```

```python
df.head()
```

```python
plt.scatter(df['Height'],df['Weight'])
plt.xlabel('Height')
plt.ylabel('Weight')
```

```python
## Divide our data set in dependent and independent variable
X=df[['Weight']]
y=df['Height']
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.20,random_state=42)
```

```python
X.shape
```

```python
X_train.shape,X_test.shape,y_train.shape,y_test.shape
```

```python
## Standardize the data
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
X_train=scaler.fit_transform(X_train)
x_test=scaler.transform(X_test)
```

```python
plt.scatter(X_train,y_train)
```

```python
## tRAIN THE MODEL simple linear regression
from sklearn.linear_model import LinearRegression
regressor=LinearRegression()
regressor.fit(X_train,y_train)
```

```python
print("The slope or coefficient is:", regressor.coef_[0])
print("The intercept is:", regressor.intercept_)
```

```python
plt.scatter(X_train,y_train)
plt.plot(X_train,regressor.predict(X_train),color='red')
```

### Prediction of train Data
1 predicted height outpur = intercept + coeef_(Weight)
2 y_pred_train = 156 + 17.03(X_train)

### Prediction of test Data
1 predicted height outpur = intercept + coeef_(Weight)
2 y_pred_train = 156 + 17.03(X_train)

```python
Y_pred=regressor.predict(X_test)
```

```python
Y_pred,y_test
```

```python
plt.scatter(X_test,y_test)
plt.plot(X_test,regressor.predict(X_test),color='red')
```

```python
## Performance METRICS

## Mean Absolute Error
## Mean Absolute Error (MAE) is the average of the absolute differences between the predicted values and the actual values. It gives an idea of how far off the predictions are from the actual values on average.
## Mean Squared Error (MSE) is the average of the squared differences between the predicted values and the actual values. It gives more weight to larger errors, making it sensitive to outliers.
## ROOT SQUARED ERROR
## R 2 Score
```

```python
from sklearn.metrics import mean_absolute_error,mean_squared_error,r2_score
print("Mean Absolute Error:", mean_absolute_error(y_test,Y_pred))
print("Mean Squared Error:", mean_squared_error(y_test,Y_pred))
print("Root Mean Squared Error:", np.sqrt(mean_squared_error(y_test,Y_pred)))
print("R 2 Score:", r2_score(y_test,Y_pred))
```
